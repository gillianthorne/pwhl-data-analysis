```python
import pandas as pd
import numpy as np
```

```python

shots = pd.read_csv("../data/shots_hometeam_2024season.csv")

shots.head()
```

```python

# sql:
# SELECT games.id, goalie.name, shots.goalie_id, shots.period, shots.x_location, shots.y_location, IF(player_teams.team_id = games.home_team_id, 1, 0) AS is_home 
# FROM shots 
# JOIN players n
# AS goalie 
# ON shots.goalie_id = goalie.id 
# JOIN games ON shots.game_id = games.id 
# JOIN player_teams 
# ON goalie.id = player_teams.player_id 
# AND player_teams.start_date != "0000-00-00"
# WHERE games.season = 1;


def do_math(x, home):
    if home == 1:
        return (600 - x)
    else:
        return x



shots["x_location"] = shots.apply(lambda x: do_math(x["x_location"], x["is_home"]), axis=1)
```

```python

# distance from net:
# let's say the net is at (585, 150)
# max x = 588, max y = 298

# a rink is 200 ft long = 2400 inches long
# therefore each unit of x is 4 inches
# a rink is 85 ft wide = 1020 inches
# therefore each unit of x is about 3.4

# distance between two points is sqrt((x1 - x2)^2 + (y1 - y2)^2) 
```

```python
shots["x_ft"] = shots["x_location"] * 4 / 12
shots["y_ft"] = shots["y_location"] * 3.4 / 12
shots["distance_ft"] = round(np.sqrt((shots["x_ft"] - 11)**2 + (shots["y_ft"] - (42.5))**2), 2)
```
